```python
import pandas as pd

# Load dataset
df = pd.read_csv(r"C:\VSCODE FOLDER\WA_Fn-UseC_-HR-Employee-Attrition.csv")

# Remove unnecessary columns
df = df.drop([
    'EmployeeCount',
    'EmployeeNumber',
    'Over18',
    'StandardHours'
], axis=1)

print("Shape after removing unnecessary columns:", df.shape)

print(df.head())
```

```python
# Check duplicate rows
duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)
```

```python
# Check missing values
missing_values = df.isnull().sum()

print(missing_values)
```

```python
# Check target distribution
print(df['Attrition'].value_counts())
```

```python
# Separate features and target

X = df.drop('Attrition', axis=1)
y = df['Attrition']

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget:")
print(y.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
print("Training target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
from imblearn.over_sampling import RandomOverSampler

# Create oversampler
ros = RandomOverSampler(random_state=42)

# Oversample ONLY training data
X_train_resampled, y_train_resampled = ros.fit_resample(
    X_train,
    y_train
)

# Check distribution
print("Before Oversampling:")
print(y_train.value_counts())

print("\nAfter Oversampling:")
print(y_train_resampled.value_counts())

print("\nNew training shape:")
print(X_train_resampled.shape)
```

```python
# Check data types
print(X_train_resampled.dtypes)
```

```python
# Numerical columns
numeric_features = X_train_resampled.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

# Categorical columns
categorical_features = X_train_resampled.select_dtypes(
    include=['object']
).columns.tolist()

print("Numerical Features:")
print(numeric_features)
print("\nTotal numerical features:", len(numeric_features))

print("\nCategorical Features:")
print(categorical_features)
print("\nTotal categorical features:", len(categorical_features))
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

# Numerical preprocessing
numeric_transformer = StandardScaler()

# Categorical preprocessing
categorical_transformer = OneHotEncoder(
    handle_unknown='ignore'
)

# Create preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

print("Preprocessor created successfully!")
```

```python
# Fit and transform training data
X_train_processed = preprocessor.fit_transform(
    X_train_resampled
)

# Transform test data
X_test_processed = preprocessor.transform(
    X_test
)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
import xgboost as xgb
```

```python
# Convert target to numeric labels

y_train_xgb = y_train_resampled.map({
    'No': 0,
    'Yes': 1
})

y_test_xgb = y_test.map({
    'No': 0,
    'Yes': 1
})

print("Training target:")
print(y_train_xgb.value_counts())

print("\nTesting target:")
print(y_test_xgb.value_counts())
```

```python
from xgboost import XGBClassifier

model_xgb = XGBClassifier(
    n_estimators=200,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric='logloss'
)

print("XGBoost model created successfully!")
```

```python
# #| Parameter          | Value | Purpose                            |
# | ------------------ | ----: | ---------------------------------- |
# | `n_estimators`     |   200 | Number of trees                    |
# | `max_depth`        |     4 | Controls tree complexity           |
# | `learning_rate`    |  0.05 | Slower learning                    |
# | `subsample`        |   0.8 | Uses 80% of training rows per tree |
# | `colsample_bytree` |   0.8 | Uses 80% of features               |
# | `random_state`     |    42 | Reproducible result                |
# Train XGBoost model

model_xgb.fit(
    X_train_processed,
    y_train_xgb
)

print("XGBoost training completed successfully!")
```

```python
# Predict on test data

y_pred_xgb = model_xgb.predict(
    X_test_processed
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)

# Predictions
y_pred_xgb = model_xgb.predict(X_test_processed)

# Probability of employee leaving
y_prob_xgb = model_xgb.predict_proba(X_test_processed)[:, 1]

# Metrics
accuracy = accuracy_score(y_test_xgb, y_pred_xgb)
precision = precision_score(y_test_xgb, y_pred_xgb)
recall = recall_score(y_test_xgb, y_pred_xgb)
f1 = f1_score(y_test_xgb, y_pred_xgb)
roc_auc = roc_auc_score(y_test_xgb, y_prob_xgb)

print("XGBoost Results")
print("=" * 40)

print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))
print("ROC-AUC  :", round(roc_auc, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_xgb, y_pred_xgb))

print("\nClassification Report:")
print(classification_report(
    y_test_xgb,
    y_pred_xgb,
    target_names=["No - Stayed", "Yes - Left"]
))
```

```python
# Training performance
y_train_pred_xgb = model_xgb.predict(X_train_processed)

train_accuracy = accuracy_score(
    y_train_xgb,
    y_train_pred_xgb
)

# Testing performance
test_accuracy = accuracy_score(
    y_test_xgb,
    y_pred_xgb
)

print("Training Accuracy:", round(train_accuracy, 4))
print("Testing Accuracy :", round(test_accuracy, 4))
print("Difference       :", round(train_accuracy - test_accuracy, 4))
```
